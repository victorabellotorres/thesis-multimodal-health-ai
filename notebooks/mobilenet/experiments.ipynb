# Nutrition5k: experimentos MobileNet

CNN móviles: convoluciones separables en profundidad, residuales invertidos y bloques diseñados con NAS.

| Clave | Modelo | Pregunta |
|---|---|---|
| `v2`, `v3`, `v4s`, `v4m` | MobileNetV2, V3-Large, V4-Conv-S, V4-Conv-M (ImageNet-1k) | Evolución de las CNN móviles (Q1) y tamaño (Q6) |
| `v4m-r256` | V4-Conv-M preentrenado a 256 px | Experimento aparte: resolución de preentreno |

- Notebook fino: solo llama a `python -m nutrition5k ...`. Toda la lógica está en `src/nutrition5k/`; los backbones
  de esta familia, en `src/nutrition5k/models/mobilenet.py`.
- Los runs se guardan en `outputs/mobilenet/<full|pilot>/<backbone>/<run>/`: `config.json` (hiperparámetros y entorno),
  `history.csv` (una fila por época), `best.pt`, `last.pt`, predicciones y métricas de validación y test.
- Funciona en local y en un pod de RunPod. En RunPod, `outputs/` se enlaza al volumen (`/workspace/nutrition5k`).
- Volver a ejecutar una celda es seguro: los runs terminados se saltan y los interrumpidos se reanudan desde `last.pt`.
- Si existe `WANDB_API_KEY`, cada run se refleja en wandb.ai (proyecto `nutrition5k`).
- Etapa 1 (L1) usa la misma configuración por defecto en las cuatro familias, así que es la comparación de
  arquitecturas en igualdad de condiciones. Etapa 2 (L2) ajusta los finalistas y la etapa 3 (L3) repite su
  configuración final con las semillas 0, 1 y 2.
- Las decisiones se toman con **validación** en `results.ipynb` de esta carpeta, nunca con test.

## Parámetros

```python
import os

FULL = True          # False: piloto 32/8 para pruebas rápidas
MODE = "complete"    # "complete" (cada 5º frame, todas las cámaras) o "minimal" (1 frame por plato)
EPOCHS = 80
BATCH = 32
WORKERS = min(15, os.cpu_count()-2)  # procesos que decodifican JPEG

FULL_FLAG = "--full" if FULL else ""
SPLIT = "full" if FULL else "pilot"
```

## Baseline (media del conjunto de entrenamiento)

```python
!{PYTHON} -m nutrition5k baseline {FULL_FLAG}
```

```python
def train(backbone, unfreeze="all", heads="single", hidden=(512,), image_size=224, backbone_lr=1e-4, seed=0,
          all_seeds=False):
    """Run one configuration; finished runs are skipped, interrupted ones resume."""
    hidden_args = " ".join(f"--hidden {size}" for size in (hidden or (0,)))
    command = (
        f"{PYTHON} -m nutrition5k train --backbone {backbone} --unfreeze {unfreeze} "
        f"--heads {heads} {hidden_args} --image-size {image_size} --backbone-lr {backbone_lr} "
        f"--epochs {EPOCHS} --batch-size {BATCH} --num-workers {WORKERS} --seed {seed} --mode {MODE} {FULL_FLAG}"
    )
    if all_seeds:
        command += " --all-seeds"
    print(command)
    get_ipython().system(command)
```

## Etapa 1 (L1): comparación de arquitecturas

Configuración por defecto, la misma en todas las familias: fine-tuning completo, cabeza `single`, capa oculta
compartida de 512, 224 px, lr del backbone 1e-4, semilla 0.

```python
for backbone in ("v2", "v3", "v4s", "v4m"):
    train(backbone)
```

## Etapa 2: ablaciones secuenciales sobre cada backbone elegido

Cada backbone de `BACKBONES` sigue su propio camino greedy: tras cada paso, mirar `results.ipynb`, elegir el ganador
**por validación** de cada backbone y escribirlo en su entrada del diccionario `BEST_*` del paso siguiente.
Si dos opciones quedan a menos de ~1 punto, es un empate: se repiten con las semillas 1 y 2 o se queda la más simple.
Los valores por defecto ya ejecutados en la etapa 1 se reutilizan, no se repiten.

### Paso 1: profundidad del fine-tuning (0 = extracción de características, N = últimas N etapas, all = completo)

```python
BACKBONES = ("v3", "v4m")  # elegidos tras la etapa 1

for backbone in BACKBONES:
    for unfreeze in (0, 1, 2):
        train(backbone, unfreeze=unfreeze)
```

### Paso 2: agrupación de cabezas (single / grouped = masa, calorías, macros / per-target)

```python
BEST_UNFREEZE = {"v3": 2, "v4m": 2}  # ganador del paso 1 (2 y all empatan; se queda 2, decisión 2026-10-23)

for backbone in BACKBONES:
    for heads in ("grouped", "per-target"):
        train(backbone, BEST_UNFREEZE[backbone], heads=heads)
```

### Paso 3: tamaño del tronco compartido

```python
BEST_HEADS = {"v3": "single", "v4m": "single"}  # ganador del paso 2 (empate de las tres; se queda single, decisión 2026-10-23)

for backbone in BACKBONES:
    for hidden in ((), (1024, 1024)):
        train(backbone, BEST_UNFREEZE[backbone], BEST_HEADS[backbone], hidden=hidden)
```

### Paso 4: resolución de entrada

```python
BEST_HIDDEN = {"v3": (512,), "v4m": ()}  # ganador del paso 3 (v3: 512; v4m: empate, se queda sin tronco, decisión 2026-10-24)

for backbone in BACKBONES:
    train(backbone, BEST_UNFREEZE[backbone], BEST_HEADS[backbone], BEST_HIDDEN[backbone], image_size=256)
```

## Etapa 3 (L3): semillas

Repetir la configuración final de cada backbone (y las que queden a menos del ruido entre semillas) con las semillas fijas 0, 1 y 2.
La semilla 0 ya está hecha y se reutiliza. `--all-seeds` escribe la media ± SD en `outputs/mobilenet/<full|pilot>/seeds/<configuración>.json`.

Además, `v4m-r256` (el mismo V4-Conv-M preentrenado a 256 px) con la configuración final de `v4m` a 256 px, en entrenamiento
y test. Es un experimento aparte, fuera del camino greedy: se compara con `v4m` a 224 px sobre las mismas tres semillas.

```python
BEST_IMAGE_SIZE = {"v3": 224, "v4m": 224}  # ganador del paso 4 (empate con 256 en ambos; se queda 224, decisión 2026-10-26)

for backbone in BACKBONES:
    train(backbone, BEST_UNFREEZE[backbone], BEST_HEADS[backbone], BEST_HIDDEN[backbone], BEST_IMAGE_SIZE[backbone],
          all_seeds=True)

# Experimento aparte: backbone preentrenado a 256 px, fine-tuning y test a 256 px.
train("v4m-r256", BEST_UNFREEZE["v4m"], BEST_HEADS["v4m"], BEST_HIDDEN["v4m"], 256, all_seeds=True)
```
